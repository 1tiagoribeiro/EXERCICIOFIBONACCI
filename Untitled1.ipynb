```python
import numpy as np
import timeit
import pandas as pd
import matplotlib.pyplot as plt
import gc

print("Bibliotecas carregadas!")
```

```python
def busca_sequencial(arr, chave):
    num_comp = 0

    for elemento in arr:
        num_comp += 1

        if elemento == chave:
            return True, num_comp

    return False, num_comp
```

```python
def busca_jump(arr, chave):
    n = len(arr)
    num_comp = 0

    if n == 0:
        return False, num_comp

    passo = int(np.sqrt(n))
    inicio = 0

    # Procura o bloco onde a chave pode estar
    while inicio < n:
        fim = min(inicio + passo, n)

        num_comp += 1
        if arr[fim - 1] >= chave:
            break
        inicio += passo

    # Busca sequencial dentro do bloco
    fim = min(inicio + passo, n)

    for i in range(inicio, fim):
        num_comp += 1

        if arr[i] == chave:
            return True, num_comp

        # Como o vetor está ordenado, podemos parar
        if arr[i] > chave:
            return False, num_comp

    return False, num_comp
```

```python
def busca_binaria(arr, chave):
    esquerda = 0
    direita = len(arr) - 1
    num_comp = 0

    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        num_comp += 1
        if arr[meio] == chave:
            return True, num_comp

        num_comp += 1
        if arr[meio] < chave:
            esquerda = meio + 1
        else:
            direita = meio - 1

    return False, num_comp

```

```python
dados = np.array([2, 5, 8, 12, 17, 23, 31, 40, 55])

print("Sequencial:", busca_sequencial(dados, 23))

print("Jump:", busca_jump(dados, 23))

print("Binária:", busca_binaria(dados, 23))
```

```python
N_values = [
    50,
    100,
    500,
    1_000,
    5_000,
    10_000,
    50_000,
    100_000,
    500_000,
    1_000_000
]

N_values
```

```python
resultados = []

for N in N_values:

    print(f"Executando N = {N:,}...")

    # Geração dos dados
    dados = np.random.randint(0, N * 10, size=N)

    # O maior valor é usado como chave
    chave = dados.max()

    # Ordenação necessária para Jump e Binária
    dados_ordenados = np.sort(dados)

    # -----------------------------
    # BUSCA SEQUENCIAL
    # -----------------------------

    inicio = timeit.default_timer()

    encontrado_seq, comp_seq = busca_sequencial(dados, chave)

    tempo_seq = timeit.default_timer() - inicio


    # -----------------------------
    # BUSCA JUMP
    # -----------------------------

    inicio = timeit.default_timer()

    encontrado_jump, comp_jump = busca_jump(dados_ordenados, chave)

    tempo_jump = timeit.default_timer() - inicio


    # -----------------------------
    # BUSCA BINÁRIA
    # -----------------------------

    inicio = timeit.default_timer()

    encontrado_bin, comp_bin = busca_binaria(dados_ordenados, chave)

    tempo_bin = timeit.default_timer() - inicio
    resultados.append({
        "N": N,
        "Tempo Sequencial (s)": tempo_seq,
        "Tempo Jump (s)": tempo_jump,
        "Tempo Binária (s)": tempo_bin,
        "Comparações Sequencial": comp_seq,
        "Comparações Jump": comp_jump,
        "Comparações Binária": comp_bin
    })

    # Liberar memória
    del dados
    del dados_ordenados
    gc.collect()


df = pd.DataFrame(resultados)

df
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    df["N"],
    df["Tempo Sequencial (s)"],
    marker="o",
    label="Busca Sequencial O(n)"
)

plt.plot(
    df["N"],
    df["Tempo Jump (s)"],
    marker="o",
    label="Busca Jump O(√n)"
)

plt.plot(
    df["N"],
    df["Tempo Binária (s)"],
    marker="o",
    label="Busca Binária O(log n)"
)

plt.xlabel("N (quantidade de elementos)")
plt.ylabel("Tempo de execução (segundos)")
plt.title("Tempo de execução em função de N")

plt.legend()
plt.grid(True)

plt.show()
```

```python
N_teste = 100_000
NUM_BUSCAS = 10_000

dados = np.random.randint(0, N_teste * 10, size=N_teste)
chave = dados.max()

inicio = timeit.default_timer()
dados_ordenados = np.sort(dados)
tempo_ordenacao = timeit.default_timer() - inicio

print(f"Tempo de ordenação: {tempo_ordenacao:.6f} segundos")

```

```python
tempo_seq_1 = timeit.timeit(
    lambda: busca_sequencial(dados, chave),
    number=1
)

tempo_jump_1 = timeit.timeit(
    lambda: busca_jump(dados_ordenados, chave),
    number=1
)

tempo_bin_1 = timeit.timeit(
    lambda: busca_binaria(dados_ordenados, chave),
    number=1
)

print(f"Sequencial: {tempo_seq_1:.6f} s")
print(f"Jump:      {tempo_jump_1:.6f} s")
print(f"Binária:   {tempo_bin_1:.6f} s")
```

```python
total_1_seq = tempo_seq_1

total_1_jump = tempo_ordenacao + tempo_jump_1

total_1_bin = tempo_ordenacao + tempo_bin_1


total_10000_seq = tempo_seq_1 * NUM_BUSCAS

total_10000_jump = tempo_ordenacao + tempo_jump_1 * NUM_BUSCAS

total_10000_bin = tempo_ordenacao + tempo_bin_1 * NUM_BUSCAS


print("===== 1 BUSCA =====")
print(f"Sequencial: {total_1_seq:.6f} s")
print(f"Jump:       {total_1_jump:.6f} s")
print(f"Binária:    {total_1_bin:.6f} s")


print("\n===== 10.000 BUSCAS =====")
print(f"Sequencial: {total_10000_seq:.6f} s")
print(f"Jump:       {total_10000_jump:.6f} s")
print(f"Binária:    {total_10000_bin:.6f} s")
```

```python
if tempo_seq_1 > tempo_bin_1:

    ponto_equilibrio_bin = (
        tempo_ordenacao /
        (tempo_seq_1 - tempo_bin_1)
    )

    print(
        f"Ponto de equilíbrio aproximado: "
        f"{ponto_equilibrio_bin:.2f} buscas"
    )

else:
    print("Não foi possível calcular o ponto de equilíbrio.")
```

```python
N_memoria = [
    1_000_000,
    2_000_000,
    5_000_000,
    10_000_000,
    20_000_000,
    50_000_000,
    100_000_000
]

resultados_memoria = []

for N in N_memoria:

    print(f"\nTestando N = {N:,}")

    try:

        dados = np.empty(N, dtype=np.int64)

        memoria_mb = dados.nbytes / (1024 ** 2)

        resultados_memoria.append({
            "N": N,
            "Memória aproximada (MB)": memoria_mb,
            "Status": "Sucesso"
        })

        print(f"Memória do array: {memoria_mb:.2f} MB")

        del dados
        gc.collect()

    except MemoryError:

        resultados_memoria.append({
            "N": N,
            "Memória aproximada (MB)": N * 8 / (1024 ** 2),
            "Status": "MemoryError"
        })

        print(">>> MEMORY ERROR <<<")

        break


df_memoria = pd.DataFrame(resultados_memoria)

df_memoria
```
